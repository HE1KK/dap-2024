Задание №1. Создать массив 10x10 со случайными значениями, найти минимум и максимум.

Ответ:

```python
import numpy as np
z = np.random.random((10, 10))
min_val = z.min()
max_val = z.max()

print("Массив 10x10:")
print(z)
print("Минимум:", min_val)
print("Максимум:", max_val)
```

Задание №2. Создать 8x8 матрицу и заполнить её единицами и нулями в шахматном порядке.

Ответ:

```python
import numpy as np
z = np.zeros((8, 8), dtype=int)
z[1::2, ::2] = 1
z[::2, 1::2] = 1

print(z)
```

Задание №3. Дан массив из 11 элементов, поменять знак у элементов, значения которых между 3 и 8.

Ответ:

```python
import numpy as np
z = np.arange(11)

mask = (z > 3) & (z < 8)
z[mask] = -z[mask]

print(z)
```

Задание №4. Найти среднее заначение строки матрицы и вычесть его из каждого елемента строки.

Ответ:

```python
import numpy as np
z = np.random.rand(5, 10)

mean_rows = z.mean(axis=1, keepdims=True)
result = z - mean_rows

print("Средние значения по строкам:")
print(mean_rows)
print("Результат:")
print(result)
```

Задание №5. Поменяйте 1 и 2 строки матрицы.

Задание №6. Дан массив 10x2 представлющий собой точки в декартовой системе координат, преобразовать в полярную.

Ответ:

```python
import numpy as np
z = np.random.random((10, 2))

x = z[:, 0]
y = z[:, 1]

r = np.sqrt(x**2 + y**2)
phi = np.arctan2(y, x)

polar = np.column_stack((r, phi))

print("Декартовы координаты (x, y):")
print(z)
print("Полярные координаты (r, phi):")
print(polar)
```

Задание №7. Заменить максимальный элемент массива на ноль

Ответ:

```python
import numpy as np
z = np.random.random(10)
print("До замены:", z)

max_index = z.argmax()
z[max_index] = 0

print("После замены:", z)
```

Задание №8. Дан массив, добавить 1 к каждому элементу с индексом, заданным в другом массиве. Добавление производиться только 1 раз.

Ответ:

```python
import numpy as np
z = np.ones(10)
i = np.random.randint(0, len(z), 20)

unique_i = np.unique(i)
z[unique_i] = z[unique_i] + 1

print("Все индексы:", i)
print("Уникальные индексы:", unique_i)
print("Итоговый массив z:", z)
```

Задание №9. Дан вектор [1, 2, 3, 4, 5], построить новый вектор с тремя нулями между каждым значением.

Ответ:

```python
import numpy as np
z = np.array([1, 2, 3, 4, 5])
zeros_count = 3

new_length = len(z) + (len(z) - 1) * zeros_count
result = np.zeros(new_length, dtype=int)
result[::zeros_count + 1] = z

print(result)
```

Задание №10. Найти ближайшее к заданному значению число в заданном массиве.

Ответ:

```python
import numpy as np
z = np.arange(100)
v = np.random.uniform(0, 100)

diff = np.abs(z - v)
closest_index = diff.argmin()
closest_value = z[closest_index]

print("Заданное число v:", v)
print("Ближайшее число в массиве:", closest_value)
```
